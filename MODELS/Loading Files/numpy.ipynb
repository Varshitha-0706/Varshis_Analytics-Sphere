**Generators**

```python
r = np.random.default_rng()
print(r)
r = np.random.default_rng(12345)
print(r)
```

```python
#Integers in the range with repetition
g = np.random.default_rng(seed=123)
g.integers(low=1, high=7, size=6)
```

```python
#Integers in the range without repetition
g = np.random.default_rng(seed=123)
g.choice(a=20, size=3, replace=False)
```

```python
# Random Shuffling
a = np.array([
[9, 4],
[3, 8],
[4, 10],
[7, 3],
[5, 10]
])
g = np.random.default_rng(seed=123)
g.permutation(a,axis=0)
```

```python
# Uniform generation as 2*3 array
ge = np.random.default_rng(seed=123)
ge.uniform(low=10.4, high=22.3, size=(2, 3))
```

```python
# Normal distribution
ge= np.random.default_rng(seed=123)
ge.normal(loc=10.0, scale=5, size=(1,2))
```

```python
# Binomial distribution
ge = np.random.default_rng(seed=123)
ge.binomial(n=10, p=0.30, size=(3,3))
```
