**ZIP**

```python
import zipfile
import pandas as pd

# open zipped dataset
with zipfile.ZipFile("csv_files.zip") as z:
    # Get a list of all CSV files within the ZIP
    csv_files = [file for file in z.namelist() if file.endswith('.csv')]

    # Choose the specific CSV file to read (e.g., the first one)
    selected_file = csv_files[0]

    # Read the selected CSV file
    with z.open(selected_file) as f:
        df = pd.read_csv(f)

# display dataset
print(df.head())
```
